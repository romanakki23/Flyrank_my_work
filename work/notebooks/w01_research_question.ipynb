## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

Search Question: Which high-exposure web pages are suffering from impression and traffic decline and should be prioritized for a content refresh?

Unit of Analysis: A single pseudonymized content page (content_id / content_hash_id).

Model Output: A probability score of decline (0.0 to 1.0) scaled into a prioritized 0–100 refresh score, paired with transparent reason codes (e.g., declining_with_demand, stale_visible_page).

Actionable Decision: The content operations team uses the top-ranked queue to assign writers to update, expand, or rewrite the top 20–50 flagged pages each month.

Cost of Being Wrong:

False Positive (ranking a healthy page high): Wasted budget and writer hours spent editing content that did not need changes.

False Negative (missing a declining page): Long-term decay in organic search visibility, leading to uncaptured traffic and lost conversions to competitors.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

Below, we load the anonymized starter dataset to extract 3 core measurements that justify Lane 2:

```python
import os
import pandas as pd

# Path resolution for both Colab and local environments
possible_paths = [
    "../../data/raw/content_refresh_anonymized.csv",
    "data/raw/content_refresh_anonymized.csv",
    "/content/flyrank/data/raw/content_refresh_anonymized.csv",
    "../data/raw/content_refresh_anonymized.csv",
]

csv_path = None
for path in possible_paths:
    if os.path.exists(path):
        csv_path = path
        break

if csv_path is None:
    # If working directly in Colab without repo clone, load directly from raw github
    csv_path = "https://raw.githubusercontent.com/romanakki23/flyrank/main/data/raw/content_refresh_anonymized.csv"

# Load dataset
df = pd.read_csv(csv_path)

# Metric 1: Total Dataset Size & Pages with Demand
total_pages = len(df)

# Metric 2: Pages showing declining trend
declining_pages = len(df[df["trend_direction"] == "down"])
decline_pct = (declining_pages / total_pages) * 100

# Metric 3: Stale High-Exposure Pages
stale_high_demand = len(
    df[(df["impressions_90d"] >= 500) & (df["content_age_days"] >= 180)]
)

print(f"1. Total anonymized pages analyzed: {total_pages:,}")
print(
    f"2. Pages in active declining trend: {declining_pages:,} ({decline_pct:.1f}% of inventory)"
)
print(
    f"3. High-exposure stale pages (>=500 imps, >=180 days old): {stale_high_demand:,}"
)
```
